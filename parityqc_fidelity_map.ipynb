```python
# --- Helper Function to compute Mean, Min, and Max per Qubit Size ---
def process_bounds(x_data, y_data):
    unique_x = np.unique(x_data)
    means, mins, maxs = [], [], []
    
    for ux in unique_x:
        # Extract all fidelity values matching this specific qubit number
        y_vals = y_data[x_data == ux]
        means.append(np.mean(y_vals))
        mins.append(np.min(y_vals))
        maxs.append(np.max(y_vals))
        
    # np.maximum(..., 1e-6) prevents log-scale clipping issues if fidelity hits 0
    return unique_x, np.array(means), np.maximum(mins, 1e-6), np.maximum(maxs, 1e-6)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
from matplotlib.ticker import LogLocator, LogFormatterMathtext

# 1. Convert raw data lists to numpy arrays (same as your current setup)
x_hh_fo = np.asarray(no_qubits_heavyhex_fo, dtype=int)
y_hh_fo = np.asarray(fidelity_heavyhex_fo, dtype=float)

x_linear_fo = np.asarray(no_qubits_linear_fo, dtype=int)
y_linear_fo = np.asarray(fidelity_linear_fo, dtype=float)

x_lf_fo = np.asarray(no_qubits_fowler_fo, dtype=int)
y_lf_fo = np.asarray(fidelity_fowler_fo, dtype=float)

x_hh = np.asarray(number_qubits_heavy_hex, dtype=int)
y_hh = np.asarray(fidelity_heavy_hex, dtype=float)

x_lf = np.asarray(number_qubits_fowler, dtype=int)
y_lf = np.asarray(fidelity_fowler, dtype=float)

x_linear = np.asarray(number_qubits_linear, dtype=int)
y_linear = np.asarray(fidelity_linear, dtype=float)


# 2. Helper function to compute Mean, Min, and Max bounds per qubit step
def process_bounds(x_data, y_data):
    unique_x = np.unique(x_data)
    means, mins, maxs = [], [], []
    for ux in unique_x:
        y_vals = y_data[x_data == ux]
        means.append(np.mean(y_vals))
        mins.append(np.min(y_vals))
        maxs.append(np.max(y_vals))
    return unique_x, np.array(means), np.array(mins), np.array(maxs)


# Process the boundaries across all 6 datasets
x_hh_fo_u, y_hh_fo_mean, y_hh_fo_min, y_hh_fo_max = process_bounds(x_hh_fo, y_hh_fo)
x_linear_fo_u, y_linear_fo_mean, y_linear_fo_min, y_linear_fo_max = process_bounds(x_linear_fo, y_linear_fo)
x_lf_fo_u, y_lf_fo_mean, y_lf_fo_min, y_lf_fo_max = process_bounds(x_lf_fo, y_lf_fo)

x_hh_u, y_hh_mean, y_hh_min, y_hh_max = process_bounds(x_hh, y_hh)
x_lf_u, y_lf_mean, y_lf_min, y_lf_max = process_bounds(x_lf, y_lf)
x_linear_u, y_linear_mean, y_linear_min, y_linear_max = process_bounds(x_linear, y_linear)

# Style color palette definitions
red = "#ff2b5f"
teal = "#65c6c4"
purple = "#6c55ff"


# 3. Structural mapping for separate independent plots
# Pairs the Fire Opal and Default lines up into separate standalone graphs
qft_configs = [
    {
        'title': 'Heavy Hex QFT Layout',
        'file_suffix': 'heavy_hex',
        'color': teal,
        'fo': (x_hh_fo_u, y_hh_fo_mean, y_hh_fo_min, y_hh_fo_max, 'Heavy Hex (Fire Opal)'),
        'df': (x_hh_u, y_hh_mean, y_hh_min, y_hh_max, 'Heavy Hex (Default)')
    },
    {
        'title': 'Linear QFT Layout',
        'file_suffix': 'linear',
        'color': red,
        'fo': (x_linear_fo_u, y_linear_fo_mean, y_linear_fo_min, y_linear_fo_max, 'Linear (Fire Opal)'),
        'df': (x_linear_u, y_linear_mean, y_linear_min, y_linear_max, 'Linear (Default)')
    },
    {
        'title': 'Fowler QFT Layout',
        'file_suffix': 'fowler',
        'color': purple,
        'fo': (x_lf_fo_u, y_lf_fo_mean, y_lf_fo_min, y_lf_fo_max, 'Fowler (Fire Opal)'),
        'df': (x_lf_u, y_lf_mean, y_lf_min, y_lf_max, 'Fowler (Default)')
    }
]


# 4. Loop over each requested qubit zoom level window
zoom_limits = [10, 20, 30, 40, 50, 60]

for max_qubits in zoom_limits:
    for config in qft_configs:
        # Generate an entirely fresh, separate full-size figure canvas
        fig, ax = plt.subplots(figsize=(4.2, 3.1), dpi=160)
        color = config['color']
        
        # Unpack variables
        x_fo, y_fo_mean, y_fo_min, y_fo_max, label_fo = config['fo']
        x_df, y_df_mean, y_df_min, y_df_max, label_df = config['df']
        
        # --- Plot curves & min-max bounds together on this plot ---
        ax.plot(x_fo, y_fo_mean, linestyle="-", marker="x", color=color, 
                linewidth=1.0, markersize=3.2, label=label_fo)
        ax.fill_between(x_fo, y_fo_min, y_fo_max, color=color, alpha=0.15, linewidth=0)

        ax.plot(x_df, y_df_mean, linestyle="--", marker="o", color=color, 
                linewidth=1.0, markersize=3.2, label=label_df)
        ax.fill_between(x_df, y_df_min, y_df_max, color=color, alpha=0.08, linewidth=0)
        
        # --- Calculate local Y bounds for this isolated view window ---
        active_y = []
        
        mask_fo = x_fo <= max_qubits
        if np.any(mask_fo):
            active_y.extend(y_fo_min[mask_fo])
            active_y.extend(y_fo_max[mask_fo])
            
        mask_df = x_df <= max_qubits
        if np.any(mask_df):
            active_y.extend(y_df_min[mask_df])
            active_y.extend(y_df_max[mask_df])
            
        ymin = min(active_y) - 0.01
        ymax = max(active_y) + 0.01
        
        # --- Format X-axis Ticks smoothly based on discrete size steps ---
        if max_qubits == 10:
            ax.xaxis.set_major_locator(ticker.MultipleLocator(2))
        elif max_qubits == 20:
            ax.xaxis.set_major_locator(ticker.MultipleLocator(4))
        elif max_qubits == 30:
            ax.xaxis.set_major_locator(ticker.MultipleLocator(5))
        else:
            ax.xaxis.set_major_locator(ticker.MultipleLocator(10))
            
        # --- Format Y-axis Ticks (Linear vs Log adaptive check) ---
        if ymin > 0.01:
            ax.set_yscale("linear")
            ax.yaxis.set_major_locator(ticker.MaxNLocator(nbins=5))
            ax.yaxis.set_major_formatter(ticker.FormatStrFormatter('%.2f'))
        else:
            ax.set_yscale("log")
            ymin = max(ymin, 1e-6)
            ax.yaxis.set_major_locator(LogLocator(base=10))
            ax.yaxis.set_major_formatter(LogFormatterMathtext(base=10))
            ax.yaxis.set_minor_locator(LogLocator(base=10, subs=np.arange(2, 10) * 0.1))

        # Apply specific window limits
        ax.set_xlim(0, max_qubits)
        ax.set_ylim(ymin, min(ymax, 1.02))
        
        # Labels and design formatting
        ax.set_title(config['title'], fontsize=8.5, fontweight='semibold', pad=8)
        ax.set_xlabel("System size (Number of Qubits)", fontsize=8)
        ax.set_ylabel("QFT Process Fidelity", fontsize=8)
        
        ax.tick_params(axis="both", which="major", labelsize=6.5, direction="in", top=True, right=True)
        ax.tick_params(axis="both", which="minor", direction="in", top=True, right=True)
        ax.grid(False)
        ax.legend(fontsize=6, frameon=False, loc="lower left", handlelength=1.8)

        plt.tight_layout(pad=0.4)
        plt.show()
```
